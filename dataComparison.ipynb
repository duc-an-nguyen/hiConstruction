```python
import pandas as pd
import os

def compare_dataframes(df1, df2):
    """
    Compare two DataFrames and print the result.

    Args:
        df1 (pd.DataFrame): The first DataFrame.
        df2 (pd.DataFrame): The second DataFrame.

    Returns:
        None: Prints whether the DataFrames are equal and any differences if found.
    """
    # Check if DataFrames are equal
    if df1.equals(df2):
        print("The DataFrames are exactly equal.")
    else:
        print("The DataFrames are not equal.")
        
        # Show differences
        diff = df1.compare(df2, keep_shape=True, keep_equal=False)
        print("Differences:")
        print(diff)

# Example usage
outputDir       = os.path.expanduser("~/Desktop/Dropbox/codeData/2ndRealDataJournalWithoutGit/")
file1 = outputDir + "data/" + "SeperatedNonCriticalData1.h5"
file2 = outputDir + "data/" + "SeperatedNonCriticalData.h5"


# Load DataFrames from HDF5 files
df1 = pd.read_hdf(file1, key='cr2')  # Replace 'dataset1' with your actual key
df2 = pd.read_hdf(file2, key='cr2')  # Replace 'dataset2' with your actual key

# Compare the DataFrames
compare_dataframes(df1, df2)
```

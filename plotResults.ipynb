```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
np.random.seed(42)
data = [np.random.normal(loc=0, scale=1, size=100),  # Dataset 1
        np.random.normal(loc=2, scale=1.5, size=100),  # Dataset 2
        np.random.normal(loc=-1, scale=0.5, size=100)]  # Dataset 3

# Create a box plot
plt.figure(figsize=(8, 6))
plt.boxplot(data, patch_artist=True, showmeans=True,
            boxprops=dict(facecolor='lightblue', color='blue'),
            meanprops=dict(marker='o', markeredgecolor='red', markerfacecolor='red'),
            medianprops=dict(color='orange'))

# Add labels and title
plt.xticks([1, 2, 3], ['Dataset 1', 'Dataset 2', 'Dataset 3'])
plt.title("Box Plot Example")
plt.ylabel("Values")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()
```

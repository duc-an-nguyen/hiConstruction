```python
import numpy as np
import matplotlib.pyplot as plt

# Polynomial function definition
def polynomial_function(x):
    return (2.83662e-7) * x**4 - 0.00000135404 * x**3 + 0.0000137975 * x**2 - 0.0247529 * x + 65.18746

# Generate x values and evaluate the polynomial function
x_values = np.linspace(0, 120, 500)  # Generate 500 points between 0 and 120
y_values = polynomial_function(x_values)

# Plotting
plt.figure(figsize=(10, 6))

plt.plot(x_values, y_values, label="y=(2.83662e-7)x^4 - 1.35404e-6x^3 + 1.37975e-5x^2 - 0.0247529x + 65.18746", color="purple")

plt.title("Polynomial Function Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()

# function 1
# if x <= 60:
#         return 65  # Constant value for x <= 60
#     else:
#         # Polynomial coefficients adjusted for continuity and smooth slope at x = 60
#         return 0.00000997592 * (x - 60)**3 + 65


# function2
#   if x <= 60:
#         return 65  # Constant value for x <= 60
#     else:
#         # Polynomial coefficients adjusted for continuity and smooth slope at x = 60
#         return 0.00005 * (x - 60)**3 + 65
```
